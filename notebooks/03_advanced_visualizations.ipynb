```python
import kagglehub
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re

path = kagglehub.dataset_download("mathurinache/list-of-countries-by-average-elevation")
csv_file = [f for f in os.listdir(path) if f.endswith('.csv')][0]
df = pd.read_csv(os.path.join(path, csv_file))


df.columns = df.columns.str.strip()
country_col = df.columns[0]
elevation_col = [c for c in df.columns if 'elevation' in c.lower()][0]


def clean_elevation(val):
    if pd.isna(val): return 0
    match = re.search(r'([-+]?[0-9,.]+)', str(val))
    if match:
        return float(match.group(1).replace(',', ''))
    return 0

df['elevation_numeric'] = df[elevation_col].apply(clean_elevation)


df_all = df.sort_values(by='elevation_numeric', ascending=True)


plt.figure(figsize=(12, 45))
sns.set_theme(style="whitegrid")


ax = sns.barplot(
    data=df_all,
    x='elevation_numeric',
    y=country_col,
    palette="viridis"
)


for i, v in enumerate(df_all['elevation_numeric']):
    ax.text(v + 5, i, f"{int(v)}m", va='center', fontsize=8)

plt.title('Tüm Ülkelerin Ortalama Yükseltileri (Sıralı Tam Liste)', fontsize=16)
plt.xlabel('Ortalama Yükselti (Metre)', fontsize=12)
plt.ylabel('Ülke', fontsize=10)


plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Veriyi hazırlama
X = yearly_temp.index.values.reshape(-1, 1)
y = yearly_temp.values

# Train-test ayrımı test=%20
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# polinom regresyon özellikleri. 2. dereceden
poly_features = PolynomialFeatures(degree=2)
X_train_poly = poly_features.fit_transform(X_train)
X_test_poly = poly_features.transform(X_test)

# Eğitim
model = LinearRegression()
model.fit(X_train_poly, y_train)

# Gelecek Tahmini
future_years = np.arange(1850, 2101).reshape(-1, 1)
future_years_poly = poly_features.transform(future_years)
future_preds = model.predict(future_years_poly)

# Rmse hesabı
y_test_pred = model.predict(X_test_poly)
test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))

# Görselleştirme
plt.figure(figsize=(15, 8))
sns.set_style("whitegrid")

plt.scatter(X_train, y_train, color='blue', alpha=0.3, s=20, label='Eğitim Verisi (%80)')
plt.scatter(X_test, y_test, color='darkblue', alpha=0.6, s=25, label='Test Verisi (%20)')

plt.plot(future_years, future_preds, color='red', linewidth=3, label='Polinom Trend (2100 Projeksiyonu)')

plt.text(0.05, 0.9, f'Test RMSE: {test_rmse:.4f} °C', transform=plt.gca().transAxes,
         fontsize=13, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, edgecolor='red'))

plt.axvline(x=X_test[0], color='black', linestyle='--', alpha=0.3) # Split çizgisi
plt.text(X_test[0], plt.ylim()[0]+0.1, ' Test Ayrımı', fontsize=10)

plt.title('Polinom Regresyon ile Küresel Sıcaklık Projeksiyonu (2100)', fontsize=16, fontweight='bold')
plt.xlabel('Yıl')
plt.ylabel('Sıcaklık (°C)')
plt.legend(loc='lower right')
plt.show()

print(f"2100 Yılı Tahmini: {future_preds[-1]:.2f} °C")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import seaborn as sns

# Veriyi hazırlama
df_xgb = yearly_temp.to_frame(name='temp')
df_xgb['year'] = df_xgb.index
df_xgb['diff'] = df_xgb['temp'].diff() # Yıllık artış miktarı

for i in range(1, 4):
    df_xgb[f'lag_{i}'] = df_xgb['diff'].shift(i)

df_xgb = df_xgb.dropna()

# Train Test ayrımı.
X = df_xgb.drop(columns=['temp', 'diff', 'year'])
y = df_xgb['diff']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# XGBoost modelini eğitme
model_xgb = xgb.XGBRegressor(n_estimators=100, learning_rate=0.05, max_depth=5)
model_xgb.fit(X_train, y_train)

# Hata hesaplama RMSE
y_pred_test = model_xgb.predict(X_test)
rmse_test = np.sqrt(mean_squared_error(y_test, y_pred_test))

# Gelecek tahmini
future_years = np.arange(df_xgb['year'].max() + 1, 2101)
last_temp = df_xgb['temp'].iloc[-1]
current_lags = df_xgb.iloc[-1][['diff', 'lag_1', 'lag_2']].values

future_preds = []
temp_acc = last_temp

for yr in future_years:
    pred_diff = model_xgb.predict(current_lags.reshape(1, -1))[0]
    temp_acc += pred_diff
    future_preds.append(temp_acc)
    current_lags = np.array([pred_diff, current_lags[0], current_lags[1]])

# Görselleştirme
plt.figure(figsize=(15, 8))
sns.set_style("whitegrid")

plt.plot(df_xgb['year'], df_xgb['temp'], color='#2c3e50', label='Gerçek Veri', alpha=0.7)

plt.plot(future_years, future_preds, color='#d35400', linewidth=3, linestyle='--', label='XGBoost 2100 Tahmini')

plt.text(0.05, 0.9, f'Test RMSE (Değişim): {rmse_test:.4f}', transform=plt.gca().transAxes,
         fontsize=12, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, edgecolor='#d35400'))

plt.title('XGBoost ile Küresel Sıcaklık Değişimi ve 2100 Projeksiyonu', fontsize=16)
plt.ylabel('Sıcaklık (°C)')
plt.legend()
plt.show()

print(f"XGBoost 2100 Yılı Tahmini: {future_preds[-1]:.2f} °C")
```

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
tf.random.set_seed(42)

# Veri Hazırlama
df_rf = yearly_temp.to_frame(name='temp')
df_rf['year'] = df_rf.index
df_rf['diff'] = df_rf['temp'].diff()

# gecikmeler
for i in range(1, 4):
    df_rf[f'lag_{i}'] = df_rf['diff'].shift(i)

df_rf = df_rf.dropna()

# Giriş ve Hedef Değişkenleri Belirleme
# LSTM'in zaman serisi mantığına tam uyması için lag_3, lag_2, lag_1 sırasıyla kronolojik veriyoruz
X = df_rf[['lag_3', 'lag_2', 'lag_1']]
y = df_rf['diff'].values.reshape(-1, 1)

X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(X, y, test_size=0.2, shuffle=False)

# ölçeklendirme
scaler_X = MinMaxScaler(feature_range=(-1, 1)) # Fark verileri eksiye düşebildiği için (-1, 1) daha kararlıdır
scaler_y = MinMaxScaler(feature_range=(-1, 1))

X_train_scaled = scaler_X.fit_transform(X_train_raw)
X_test_scaled = scaler_X.transform(X_test_raw)

y_train_scaled = scaler_y.fit_transform(y_train_raw)
y_test_scaled = scaler_y.transform(y_test_raw)


# 3 zaman adımı (lag_3 -> lag_2 -> lag_1)
X_train_lstm = X_train_scaled.reshape((X_train_scaled.shape[0], 3, 1))
X_test_lstm = X_test_scaled.reshape((X_test_scaled.shape[0], 3, 1))

# Modeli Optimize Etme
model_lstm = Sequential([
    LSTM(16, activation='tanh', input_shape=(3, 1), return_sequences=False),
    Dropout(0.1),
    Dense(8, activation='relu'),
    Dense(1)
])

# Öğrenme oranı 0.005
model_lstm.compile(optimizer=Adam(learning_rate=0.005), loss='mse')

# Modeli Eğitme
model_lstm.fit(
    X_train_lstm, y_train_scaled,
    epochs=120,
    batch_size=8,
    validation_data=(X_test_lstm, y_test_scaled),
    verbose=0
)

# Hata Hesaplama
y_pred_scaled = model_lstm.predict(X_test_lstm)
y_pred_test = scaler_y.inverse_transform(y_pred_scaled)

rmse_test = np.sqrt(mean_squared_error(y_test_raw, y_pred_test))

# Gelecek Tahmini
future_years = np.arange(df_rf['year'].max() + 1, 2101)
last_temp = df_rf['temp'].iloc[-1]

current_lags_raw = df_rf.iloc[-1][['lag_3', 'lag_2', 'lag_1']].values.reshape(1, -1)
current_lags_scaled = scaler_X.transform(current_lags_raw)

future_preds = []
temp_acc = last_temp

for yr in future_years:
    # Formatı [1, 3, 1] yapıyoruz
    current_lstm_input = current_lags_scaled.reshape((1, 3, 1))

    pred_diff_scaled = model_lstm.predict(current_lstm_input, verbose=0)[0]
    pred_diff = scaler_y.inverse_transform(pred_diff_scaled.reshape(-1, 1))[0][0]

    temp_acc += pred_diff
    future_preds.append(temp_acc)


    new_lags_raw = np.array([current_lags_raw[0][1], current_lags_raw[0][2], pred_diff]).reshape(1, -1)
    current_lags_scaled = scaler_X.transform(new_lags_raw)
    current_lags_raw = new_lags_raw

# Görselleştirme
plt.figure(figsize=(15, 8))
sns.set_style("whitegrid")

plt.plot(df_rf['year'], df_rf['temp'], color='#2980b9', label='Gerçek Veri (Geçmiş)', alpha=0.7)
plt.plot(future_years, future_preds, color='#e67e22', linewidth=3, linestyle='--', label='LSTM 2100 Tahmini')

plt.text(0.05, 0.9, f'Test RMSE (Değişim): {rmse_test:.4f}', transform=plt.gca().transAxes,
         fontsize=12, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, edgecolor='#e67e22'))

plt.title('LSTM ile Küresel Sıcaklık Analizi (2100)', fontsize=16)
plt.ylabel('Sıcaklık (°C)')
plt.legend()
plt.show()

print(f"Yeni LSTM 2100 Yılı Tahmini: {future_preds[-1]:.2f} °C")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Veri Hazırlama
full_path = os.path.join(path, 'Global_sea_level_rise.csv')
df_sea = pd.read_csv(full_path)
df_sea['date'] = pd.to_datetime(df_sea['date'])
df_sea['year'] = df_sea['date'].dt.year
yearly_sea = df_sea.groupby('year')['mmfrom1993-2008average'].mean().to_frame(name='level')

df_xgb = yearly_sea.copy()
df_xgb['diff'] = df_xgb['level'].diff()

for i in range(1, 4):
    df_xgb[f'lag_{i}'] = df_xgb['diff'].shift(i)

df_xgb = df_xgb.dropna()

# Train-Test ayrımı
X = df_xgb.drop(columns=['level', 'diff'])
y = df_xgb['diff']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# XGBoost modelini eğitme
model_xgb = xgb.XGBRegressor(n_estimators=150, learning_rate=0.05, max_depth=4)
model_xgb.fit(X_train, y_train)

# Gelecek Tahmini
future_years = np.arange(df_xgb.index.max() + 1, 2101)
last_real_level = df_xgb['level'].iloc[-1]
current_lags = df_xgb.iloc[-1][['diff', 'lag_1', 'lag_2']].values

future_preds = []
level_accumulator = last_real_level

for yr in future_years:
    pred_diff = model_xgb.predict(current_lags.reshape(1, -1))[0]
    level_accumulator += pred_diff
    future_preds.append(level_accumulator)

    current_lags = np.array([pred_diff, current_lags[0], current_lags[1]])

# Görselleştirme
plt.figure(figsize=(15, 8))
sns.set_style("whitegrid")

plt.plot(yearly_sea.index, yearly_sea['level'], color='#2c3e50', label='Gerçek Veri', alpha=0.7)

plt.plot(future_years, future_preds, color='#e67e22', linewidth=3, linestyle='--', label='XGBoost 2100 Tahmini')

test_preds_diff = model_xgb.predict(X_test)
rmse_test = np.sqrt(mean_squared_error(y_test, test_preds_diff))

plt.text(0.05, 0.9, f'Test RMSE (Değişim): {rmse_test:.4f} mm', transform=plt.gca().transAxes,
         fontsize=12, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, edgecolor='#e67e22'))

plt.title('XGBoost ile Deniz Seviyesi Artış Projeksiyonu (2100)', fontsize=16)
plt.ylabel('Deniz Seviyesi Değişimi (mm)')
plt.legend()
plt.show()

print(f"XGBoost 2100 Yılı Tahmini: {future_preds[-1]:.2f} mm")
```

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Tekrarlanabilirlik için sabit tohum (seed) değerleri
np.random.seed(42)
tf.random.set_seed(42)

# Veriyi Hazırlama
full_path = os.path.join(path, 'Global_sea_level_rise.csv')
df_sea = pd.read_csv(full_path)
df_sea['date'] = pd.to_datetime(df_sea['date'])
df_sea['year'] = df_sea['date'].dt.year
yearly_sea = df_sea.groupby('year')['mmfrom1993-2008average'].mean().to_frame(name='level')

df_rf = yearly_sea.copy()
df_rf['diff'] = df_rf['level'].diff()

for i in range(1, 4):
    df_rf[f'lag_{i}'] = df_rf['diff'].shift(i)

df_rf = df_rf.dropna()

# Train Test ayrımı
X = df_rf.drop(columns=['level', 'diff'])
y = df_rf['diff'].values.reshape(-1, 1)

X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(X, y, test_size=0.2, shuffle=False)

# LSTM için Ölçeklendirme (Scaling)
scaler_X = MinMaxScaler(feature_range=(0, 1))
scaler_y = MinMaxScaler(feature_range=(0, 1))

X_train_scaled = scaler_X.fit_transform(X_train_raw)
X_test_scaled = scaler_X.transform(X_test_raw)

y_train_scaled = scaler_y.fit_transform(y_train_raw)
y_test_scaled = scaler_y.transform(y_test_raw)

# LSTM Veri Formatı Düzenleme [samples, time_steps, features]
X_train_lstm = X_train_scaled.reshape((X_train_scaled.shape[0], 1, X_train_scaled.shape[1]))
X_test_lstm = X_test_scaled.reshape((X_test_scaled.shape[0], 1, X_test_scaled.shape[1]))

# LSTM Modelini Kurma
model_lstm_sea = Sequential([
    LSTM(64, activation='tanh', input_shape=(X_train_lstm.shape[1], X_train_lstm.shape[2]), return_sequences=False),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1) # çıkış katmanı
])

model_lstm_sea.compile(optimizer='adam', loss='mse')

# Modeli Eğitme
model_lstm_sea.fit(
    X_train_lstm, y_train_scaled,
    epochs=100,
    batch_size=4,
    validation_data=(X_test_lstm, y_test_scaled),
    verbose=0
)

# Hata Hesaplama
y_pred_scaled = model_lstm_sea.predict(X_test_lstm)
y_pred_test = scaler_y.inverse_transform(y_pred_scaled)

rmse_test = np.sqrt(mean_squared_error(y_test_raw, y_pred_test))

# Gelecek Tahmini (Özyinelemeli / Recursive Forecasting)
future_years = np.arange(df_rf.index.max() + 1, 2101)
last_real_level = df_rf['level'].iloc[-1]

# Son 3 yılın fark değerlerini alıp ölçeklendirme
current_lags_raw = df_rf.iloc[-1][['diff', 'lag_1', 'lag_2']].values.reshape(1, -1)
current_lags_scaled = scaler_X.transform(current_lags_raw)

future_preds = []
level_acc = last_real_level

for yr in future_years:

    current_lstm_input = current_lags_scaled.reshape((1, 1, 3))


    pred_diff_scaled = model_lstm_sea.predict(current_lstm_input, verbose=0)[0]


    pred_diff = scaler_y.inverse_transform(pred_diff_scaled.reshape(-1, 1))[0][0]


    level_acc += pred_diff
    future_preds.append(level_acc)

    # Lag'leri bir sonraki adım için güncelle ve ölçeklendir
    new_lags_raw = np.array([pred_diff, current_lags_raw[0][0], current_lags_raw[0][1]]).reshape(1, -1)
    current_lags_scaled = scaler_X.transform(new_lags_raw)
    current_lags_raw = new_lags_raw

# Görselleştirme
plt.figure(figsize=(15, 8))
sns.set_style("whitegrid")

plt.plot(yearly_sea.index, yearly_sea['level'], color='#34495e', label='Gerçek Veri', alpha=0.7)
plt.plot(future_years, future_preds, color='#e67e22', linewidth=3, linestyle='--', label='LSTM 2100 Tahmini')

plt.text(0.05, 0.9, f'Test RMSE (Değişim): {rmse_test:.4f} mm', transform=plt.gca().transAxes,
         fontsize=12, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8, edgecolor='#e67e22'))

plt.title('LSTM ile Deniz Seviyesi Artış Analizi (2100)', fontsize=16)
plt.ylabel('Deniz Seviyesi Değişimi (mm)')
plt.legend()
plt.show()

print(f"LSTM 2100 Yılı Tahmini: {future_preds[-1]:.2f} mm")
```
